```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
import matplotlib.colors as mcolors

f_size = 12

data_true = np.load("./simulation/data_0.25/mv_sim_seed0.npz")
x_true = data_true["x_paths"]

lambda_list = [
    ("true",                 "True"),
    ("lambda_001_0p000e00",  "$\\lambda=0$"),
    ("lambda_004_1p000e-05", "$\\lambda=10^{-5}$"),
    ("lambda_006_1p000e-03", "$\\lambda=10^{-3}$"),
]

nudge = 10
x_grid = np.linspace(-1, 1, 200)

# compute all kde_time arrays first to get global vmin/vmax
kde_all = []
for lam_str, _ in lambda_list:
    if lam_str == "true":
        x = x_true
    else:
        path = f"./assimilation/data_0.25/lambda_scan_parameter_grid/a_1p000e-01__nudge_iters_{nudge:03d}/lambda_scan/runs/{lam_str}/x_da_kdenudge_1d_seed0.npz"
        x = np.load(path)["x_est"]
    kde_time = np.array([gaussian_kde(x[t, :])(x_grid) for t in range(x.shape[0])])
    kde_all.append(kde_time)

vmin = min(k.min() for k in kde_all)
vmax = max(k.max() for k in kde_all)

T = kde_all[0].shape[0]
dt = 0.01
time_max = (T - 1) * dt

fig, axes = plt.subplots(
    2, 2, figsize=(10, 8),
    sharex=True, sharey=True
)

axes = axes.flatten()
fig.subplots_adjust(wspace=0, hspace=0)

levels = 16  # number of color bins
norm = mcolors.BoundaryNorm(
    np.linspace(vmin, vmax, levels),
    ncolors=plt.cm.viridis.N
)
pad = 0.1  # small extension

for idx, (kde_time, (_, lam_label)) in enumerate(zip(kde_all, lambda_list)):
    ax = axes[idx]
    im = ax.imshow(
        kde_time,
        aspect='auto',
        origin='lower',
        extent=[x_grid[0]-pad, x_grid[-1]+pad, 0, time_max],
        norm=norm,
        interpolation='nearest'
    )

    ax.text(
        0.5, 0.98, lam_label,
        transform=ax.transAxes,
        ha='center', va='top', fontsize=f_size
    )

    if idx in (0, 2):
        ax.set_ylabel("time", fontsize=f_size)
    if idx in (2, 3):
        ax.set_xlabel("x", fontsize=f_size)

# colorbar
cbar = fig.colorbar(im, ax=axes, label="density", shrink=0.6)
ticks = np.linspace(0.2,0.8,4)  # choose how many ticks you want
cbar.set_ticks(ticks)
cbar.ax.tick_params(labelsize=f_size)

for ax in axes:
    ax.set_yticks([0,1.0,2.0,3.0,4.0])
    ax.tick_params(axis='y', labelsize=f_size)
plt.savefig("case2_density_true_0.25_a_0.1.pdf", bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ot

data_true = np.load("./simulation/data_0.25/mv_sim_seed0.npz")
x_true = data_true["x_paths"]

lambda_list = [
    ("lambda_001_0p000e00",  0.0),
    ("lambda_002_1p000e-07", 1e-7),
    ("lambda_003_1p000e-06", 1e-6),
    ("lambda_004_1p000e-05", 1e-5),
    ("lambda_005_1p000e-04", 1e-4),
    ("lambda_006_1p000e-03", 1e-3)
]

a_list = [0.1,0.5,1.5]
nudge = 20

def wasserstein2(x_est, x_true):
    x_est = np.asarray(x_est).reshape(-1, 1)
    x_true = np.asarray(x_true).reshape(-1, 1)

    a = np.ones(x_est.shape[0]) / x_est.shape[0]
    b = np.ones(x_true.shape[0]) / x_true.shape[0]

    M = ot.dist(x_est, x_true, metric="sqeuclidean")

    return np.sqrt(ot.emd2(a, b, M))


results = {a: [] for a in a_list}

x_true_T = x_true[-1]

def format_a(a):
    s = f"{a:.3e}"          # e.g. 5.000e-01 or 2.000e+00
    s = s.replace(".", "p")
    s = s.replace("e+", "e")  # remove +
    return s

for a in a_list:
    for lam_str, lam_val in lambda_list:
        a_str = format_a(a)
        path = (f"./assimilation/data_0.25/lambda_scan_parameter_grid/a_{a_str}__nudge_iters_{nudge:03d}/lambda_scan/runs/\{lam_str}/x_da_kdenudge_1d_seed0.npz")
        data = np.load(path)
        x_est = data["x_est"]
        x_est_T = x_est[-1]

        w2 = wasserstein2(x_est_T, x_true_T)
        results[a].append(w2)


plt.figure()
for a in a_list:
    plt.plot(
        [lam for _, lam in lambda_list],
        results[a],
        marker='o',
        label=f"a = {a}"
    )

plt.xscale("log")
plt.xlabel(r"$\lambda$")
plt.ylabel(r"final-time $W_2$ error")
plt.legend()
plt.grid(True, alpha=0.3)

plt.savefig("case2_true_0.25_finaltime_w2.pdf", bbox_inches="tight", pad_inches=0.05)
plt.show()
```
